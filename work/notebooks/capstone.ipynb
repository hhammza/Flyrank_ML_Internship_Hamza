### Five-minute demo outline

**0:00?0:40 ? The question.** FlyRank content teams have more pages than they can review. I ask which visible pages look worth checking first because they receive fewer clicks than their search visibility suggests.

**0:40?1:30 ? The data and lane.** I use the March 2026 partition of the FlyRank warehouse. One row is a client-content item on a reporting date. The decision-time features are impressions, clicks, average position, and CTR; identifiers are grouping keys only.

**1:30?2:30 ? The method.** I compare the Week-4 transparent rule with Logistic Regression. The rule selects pages with at least 500 impressions, position 4?20, and CTR below 0.5%. The model is evaluated with a client-grouped holdout so unseen-client performance is the honest headline.

**2:30?3:30 ? One chart.** Show the model-versus-baseline chart: ROC-AUC, average precision, and precision@50. Explain the base rate beside the metrics and say that the comparison is measured on this March slice, not a universal guarantee.

**3:30?4:20 ? One honest result.** The model measured ROC-AUC 0.8458 and average precision 0.5224 versus 0.5000 and 0.1386 for the rule; precision@50 was 0.94 for the model versus 0.08 for the rule. Errors remain because next-day search activity is noisy.

**4:20?5:00 ? One recommendation and boundary.** Start human review with high-reach, low-CTR pages. Do not auto-publish edits, change technical SEO controls, delete content, or claim that an edit caused growth.
